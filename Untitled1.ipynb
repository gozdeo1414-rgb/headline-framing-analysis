```python
data = {
    "id": list(range(1, 26)),

    "topic": (
        ["Environment"] * 5 +
        ["Artificial Intelligence"] * 5 +
        ["Economy"] * 5 +
        ["Health"] * 5 +
        ["Technology"] * 5
    ),

    "framing": (
        ["Neutral", "Alarmist", "Positive", "Curiosity", "Sensational"] * 5
    ),

    "headline": [
        # Environment
        "New research examines the health effects of air pollution in cities",
        "Experts warn that air pollution in cities may pose a serious health threat",
        "New research offers promising insights into the fight against urban air pollution",
        "What could new research reveal about the air we breathe in cities?",
        "The hidden danger in city air that researchers are now uncovering",

        # Artificial Intelligence
        "Researchers examine how artificial intelligence is changing workplace tasks",
        "AI could dramatically transform the way millions of people work, researchers warn",
        "New AI developments could create new opportunities for workers",
        "What will artificial intelligence change about the way we work?",
        "The AI revolution is coming for the workplace — and few may be ready",

        # Economy
        "New data shows changes in household spending patterns",
        "Rising household costs are putting increasing pressure on consumers",
        "New spending trends may offer signs of changing consumer behavior",
        "What do recent spending patterns reveal about households?",
        "The spending numbers that could change how we see the economy",

        # Health
        "Researchers study the relationship between sleep and cognitive performance",
        "Lack of sleep may be putting people's cognitive health at risk",
        "New research provides encouraging insights into the benefits of healthy sleep",
        "Could the way you sleep affect how your brain performs?",
        "What scientists discovered about sleep could change your daily routine",

        # Technology
        "Researchers examine how social media recommendation systems affect users",
        "Social media algorithms may be influencing what millions of users see",
        "New research highlights potential benefits of more transparent recommendation systems",
        "How much do social media algorithms really shape what we see?",
        "The algorithm deciding what you see may know more about you than you think"
    ]
}

df = pd.DataFrame(data)

df
```

```python
df["word_count"] = df["headline"].str.split().str.len()

df[["topic", "framing", "headline", "word_count"]]
```

```python
df["question_mark"] = df["headline"].str.contains(r"\?").astype(int)
df["exclamation_mark"] = df["headline"].str.contains(r"!").astype(int)

df[["framing", "headline", "question_mark", "exclamation_mark"]]
```

```python
positive_words = [
    "promising", "benefits", "encouraging", "opportunities",
    "hope", "positive", "improve", "improvement"
]

negative_words = [
    "danger", "dangerous", "threat", "risk", "warning",
    "warn", "pressure", "problem", "harm", "hidden"
]
```

```python
def count_words(text, word_list):
    words = text.lower().split()
    return sum(word.strip(".,!?;:") in word_list for word in words)

df["positive_word_count"] = df["headline"].apply(
    lambda x: count_words(x, positive_words)
)

df["negative_word_count"] = df["headline"].apply(
    lambda x: count_words(x, negative_words)
)

df[[
    "framing",
    "headline",
    "positive_word_count",
    "negative_word_count"
]]
```

```python
framing_summary = df.groupby("framing")[
    ["word_count", "positive_word_count", "negative_word_count",
     "question_mark", "exclamation_mark"]
].mean()

framing_summary
```

```python
df.groupby("framing")
```

```python
framing_summary = df.groupby("framing")[
    ["word_count", "positive_word_count", "negative_word_count",
     "question_mark", "exclamation_mark"]
].mean()

framing_summary
```

```python
import matplotlib.pyplot as plt

framing_summary["word_count"].plot(kind="bar")

plt.title("Average Headline Length by Framing")
plt.xlabel("Framing")
plt.ylabel("Average Number of Words")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
framing_summary[[
    "positive_word_count",
    "negative_word_count"
]].plot(kind="bar")

plt.title("Positive and Negative Language by Framing")
plt.xlabel("Framing")
plt.ylabel("Average Word Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
question_summary = df.groupby("framing")["question_mark"].sum()

question_summary
```

```python
question_summary.plot(kind="bar")

plt.title("Question Mark Usage by Framing")
plt.xlabel("Framing")
plt.ylabel("Number of Headlines")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["average_word_length"] = df["headline"].apply(
    lambda x: sum(len(word.strip(".,!?;:")) for word in x.split()) / len(x.split())
)

df[[
    "framing",
    "headline",
    "average_word_length"
]]
```

```python
average_word_length_summary = df.groupby("framing")[
    "average_word_length"
].mean().sort_values(ascending=False)

average_word_length_summary.round(2)
```

```python
average_word_length_summary.plot(kind="bar")

plt.title("Average Word Length by Framing")
plt.xlabel("Framing")
plt.ylabel("Average Word Length")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from textblob import TextBlob
```

```python
df["sentiment_score"] = df["headline"].apply(
    lambda x: TextBlob(x).sentiment.polarity
)

df[[
    "framing",
    "headline",
    "sentiment_score"
]]
```

```python
sentiment_summary = df.groupby("framing")[
    "sentiment_score"
].mean().sort_values(ascending=False)

sentiment_summary.round(3)
```

```python
sentiment_summary.plot(kind="bar")

plt.title("Average Sentiment Score by Framing")
plt.xlabel("Framing")
plt.ylabel("Average Sentiment Score")
plt.axhline(0, linewidth=1)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sentiment_range = df.groupby("framing")["sentiment_score"].agg(
    ["mean", "min", "max"]
)

sentiment_range.round(3)
```
